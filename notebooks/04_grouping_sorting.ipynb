```python
import pandas as pd

columns = ['engine_id', 'cycle',
           'op_setting1', 'op_setting2', 'op_setting3'] + \
          [f'sensor_{i}' for i in range(1, 22)]

train = pd.read_csv('../datasets/train_FD001.txt',
                    sep=r'\s+',
                    header=None)

train = train.dropna(axis=1)
train.columns = columns
# Sort by cycle (ascending)
train.sort_values(by='cycle').head()

# Sort by cycle (descending)
train.sort_values(by='cycle', ascending=False).head()

# Sort by engine_id and cycle
train.sort_values(by=['engine_id', 'cycle']).head()
# Get max cycle per engine and sort by highest life
max_cycles = train.groupby('engine_id')['cycle'].max()

max_cycles.sort_values(ascending=False).head()
train.groupby('engine_id').agg({
    'cycle': 'max',
    'sensor_1': 'mean',
    'sensor_2': 'mean'
}).head()
```
