```python
# 01_python.ipynb
# Basic loading and exploration of NASA CMAPSS dataset

import pandas as pd

# Define column names for NASA CMAPSS dataset
columns = ['engine_id', 'cycle',
           'op_setting1', 'op_setting2', 'op_setting3'] + \
          [f'sensor_{i}' for i in range(1, 22)]

# Load the dataset
train = pd.read_csv(
    '../datasets/train_FD001.txt',   # change path if needed
    sep=r'\s+',
    header=None
)

# Remove extra blank columns
train = train.dropna(axis=1)

# Assign column names
train.columns = columns

# Display first 5 rows
train.head()
# Display last 5 rows
train.tail()

# Shape of dataset
train.shape

# Column names
train.columns

# Basic statistical summary
train.describe()
```
