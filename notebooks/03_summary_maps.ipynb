```python
import pandas as pd

columns = ['engine_id', 'cycle',
           'op_setting1', 'op_setting2', 'op_setting3'] + \
          [f'sensor_{i}' for i in range(1, 22)]

train = pd.read_csv('../datasets/train_FD001.txt',
                    sep=r'\s+',
                    header=None)

train = train.dropna(axis=1)
train.columns = columns
# Overall summary
train.describe()

# Mean of all columns
train.mean()

# Max and Min of sensor_1
train['sensor_1'].max()
train['sensor_1'].min()

# Median of cycle
train['cycle'].median()

# Standard deviation of sensor_5
train['sensor_5'].std()
# Maximum cycle per engine
max_cycles = train.groupby('engine_id')['cycle'].max()
max_cycles.head()

# Average sensor values per engine
mean_sensors = train.groupby('engine_id').mean()
mean_sensors.head()

# Number of cycles per engine
cycle_count = train.groupby('engine_id')['cycle'].count()
cycle_count.head()
```
